```python
from youtube_transcript_api import YouTubeTranscriptApi

video_id = "wwSzpaTHyS8"

from langchain_community.document_loaders import YoutubeLoader

loader = YoutubeLoader.from_youtube_url(f"https://www.youtube.com/watch?v={video_id}", add_video_info=False)
transcript = loader.load()
```

```python
transcript
```

```python
text = transcript[0].page_content

text = text.replace('\xa0', ' ')

text = ' '.join(text.split())

print(text[:500])
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = splitter.create_documents([text])
```

```python
print(len(chunks))
print(len(chunks[1].page_content))
chunks[13]
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
vector_store = FAISS.from_documents(chunks, embeddings)
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(['51658a75-3ea9-4522-9649-d450aa5feda8'])
```

```python
retriever = vector_store.as_retriever(search_type="similarity", search_kwargs={"k": 4})
retriever
```

```python
retriever.invoke('What is Time')
```

```python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.2)

prompt = PromptTemplate(
    template="""
      You are a helpful assistant.
      Answer only and only from the provided transcript context.
      If the context is insufficient, just say you don't know.

      {context}
      Question: {question}
    """,
    input_variables = ['context', 'question']
)
```

```python
question          = "is time relative ? if yes then what was discussed"
retrieved_docs    = retriever.invoke(question)
retrieved_docs
```

```python
context_text = '\n\n'.join(doc.page_content for doc in retrieved_docs)
context_text
```

```python
final_prompt = prompt.invoke({'context': context_text, "question": question})
answer = llm.invoke(final_prompt)
print(answer.content)
```
