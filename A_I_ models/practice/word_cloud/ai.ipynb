```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = """
artificial intelligence machine learning deep learning data science
neural network robotics automation technology cyber digital future
algorithm computer vision natural language processing ai innovation
"""

wc = WordCloud(
    width=1200,
    height=600,
    background_color="white",
    colormap="Dark2",
    max_words=200
).generate(text)

plt.figure(figsize=(12,6))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```

```python
import matplotlib.pyplot as plt
import random

from wordcloud import WordCloud

# Existing high-frequency terms
frequencies = {
    "Artificial Intelligence": 120,
    "Machine Learning": 110,
    "Deep Learning": 100,
    "Data Science": 90,
    "Neural Networks": 85,
    "Computer Vision": 80,
    "Natural Language Processing": 70,
    "Optimization": 60,
    "Model Training": 55,
    "Feature Engineering": 50,
    "MLOps": 45,
    "Research": 40,
    "Algorithms": 38,
    "Statistics": 35,
    "Cloud Computing": 30,
    "GPU": 58
}

# New terms with lower frequencies
new_terms = [
    "model evaluation", "generalization", "loss functions",
    "data analytics", "data visualization", "exploratory analysis",
    "statistical modeling", "probability", "hypothesis testing",
    "regression", "classification", "clustering", "dimensionality reduction",
    "sentiment analysis", "transformers", "attention mechanism",
    "language models", "image recognition", "object detection",
    "semantic segmentation", "generative models", "diffusion models",
    "reinforcement learning", "autonomous systems", "roboticsAPI", "pipeline automation"
]

# Assign lower frequencies
for term in new_terms:
    frequencies[term] = 30  # low frequency for new terms

# Generate WordCloud
wc = WordCloud(
    width=2000,    # width in pixels
    height=500,    # height in pixels
    background_color="#f8fafc",
    colormap="Blues",
    max_words=500,
    prefer_horizontal=0.65,
    relative_scaling=0.4,
    font_step=2
).generate_from_frequencies(frequencies)

# Display
plt.figure(figsize=(20, 5))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()

# Save as PNG with random 3-digit number
random_number = random.randint(100, 999)
filename = f"wordcloud_{random_number}.png"
wc.to_file(filename)
print(f"Word cloud saved as {filename}")
```

```python
from PIL import Image, ImageOps

# Path to your original image
image_path = "wordcloud_521.png"  # replace with your image path
background_color = "#f8fafc"
# Open original image
img = Image.open(image_path)
orig_width, orig_height = img.size

# Calculate new dimensions with margins
new_width = int(orig_width * 1.3 * 2 / 1.0)  # 30% margin on each side
new_height = int(orig_height * 1.5 * 2 / 1.0)  # 50% margin on top and bottom

# Actually, simpler: add 30% width each side, 50% height top/bottom
margin_left_right = int(orig_width * 0.4)
margin_top_bottom = int(orig_height * 0.9)

new_width = orig_width + 2 * margin_left_right
new_height = orig_height + 2 * margin_top_bottom

# Create new image with background color
new_img = Image.new("RGB", (new_width, new_height), color=background_color)

# Paste original image in the center
new_img.paste(img, (margin_left_right, margin_top_bottom))

# Save or show the new image
new_img.save("wordcloud_with_margins.png")
new_img.show()
```
